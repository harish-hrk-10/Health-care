```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/healthcare_preprocessed")
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print(df.columns.tolist())
```

```python
df.info()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x='Age',
    bins=15,
    kde=True
)

plt.title('Distribution of Patient Age')
plt.xlabel('Age')
plt.ylabel('Number of Patients')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Age',
    hue='Gender',
    bins=15,
    kde=True,
    element='step'
)

plt.title('Patient Age Distribution by Gender')
plt.xlabel('Age')
plt.ylabel('Number of Patients')

plt.show()
```

```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(7, 5))

sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values
)

plt.title('Patient Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Patients')

plt.show()
```

```python
condition_counts = df['Medical_Condition'].value_counts()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=condition_counts.values,
    y=condition_counts.index
)

plt.title('Number of Patients by Medical Condition')
plt.xlabel('Number of Patients')
plt.ylabel('Medical Condition')
plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x='Medical_Condition',
    hue='Gender'
)

plt.title('Gender Distribution Across Medical Conditions')
plt.xlabel('Medical Condition')
plt.ylabel('Number of Patients')

plt.xticks(rotation=10)

plt.show()
```

```python
admissions_counts = df['Admission_Type'].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=admissions_counts.index,
    y=admissions_counts.values
)

plt.title('Distribution of Admission Type ')
plt.xlabel('Admission Type')
plt.ylabel('Number of Patients')

plt.show()
```

```python
admissions_counts = pd.crosstab(
    df['Medical_Condition'],
    df['Admission_Type']
)

print(admissions_counts)
```

```python
admissions_counts.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Admission Type Across Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Number of Patients')

plt.xticks(rotation=30)

plt.show()
```
